```python
from sklearn.feature_extraction.text import CountVectorizer

# Fungsi untuk ekstrak N-gram dan tampilkan top frekuensi
def analyze_ngrams(texts, ngram_range, ngram_name, top_n=15):
    vectorizer = CountVectorizer(ngram_range=ngram_range, max_features=1000, lowercase=True)
    X = vectorizer.fit_transform(texts)
    
    # Hitung frekuensi
    word_freq = dict(zip(vectorizer.get_feature_names_out(), X.toarray().sum(axis=0)))
    word_freq_sorted = sorted(word_freq.items(), key=lambda x: x[1], reverse=True)
    
    # Tampilkan top N
    print(f"\n{'='*60}")
    print(f"Top {top_n} {ngram_name} (Paling Sering Muncul)")
    print(f"{'='*60}")
    for i, (word, freq) in enumerate(word_freq_sorted[:top_n], 1):
        print(f"{i:2d}. {word:30s} : {freq:5d} kali")
    
    # Visualisasi
    top_words = word_freq_sorted[:top_n]
    words = [item[0] for item in top_words]
    freqs = [item[1] for item in top_words]
    
    plt.figure(figsize=(12, 6))
    plt.barh(range(len(words)), freqs, color='teal', alpha=0.7)
    plt.yticks(range(len(words)), words)
    plt.xlabel('Frekuensi')
    plt.title(f'Top {top_n} {ngram_name} dalam Data Train')
    plt.gca().invert_yaxis()
    plt.tight_layout()
    plt.show()

analyze_ngrams(train['text'], (1, 1), "UNIGRAM")
analyze_ngrams(train['text'], (2, 2), "BIGRAM")
analyze_ngrams(train['text'], (3, 3), "TRIGRAM")
```

```python
# 3. IDENTIFIKASI ENTITAS (ENTITY MASKING)
def extract_entities(text):
    """Ekstrak entitas yang ditandai dengan pola {@...@}"""
    entities = re.findall(r'\{@([^@]+)@\}', text)
    return entities

# Ekstrak semua entitas dari train dan test
train_entities = []
for idx, text in enumerate(train['text']):
    entities = extract_entities(text)
    train_entities.extend(entities)

test_entities = []
for idx, text in enumerate(test['text']):
    entities = extract_entities(text)
    test_entities.extend(entities)

# Hitung frekuensi entitas
from collections import Counter

train_entity_freq = Counter(train_entities)
test_entity_freq = Counter(test_entities)

print("="*70)
print("ENTITY MASKING ANALYSIS: Pola {@...@}")
print("="*70)
print(f"\nTotal Entitas dalam Train: {len(train_entities)}")
print(f"Total Entitas Unik dalam Train: {len(train_entity_freq)}")
print(f"\nTotal Entitas dalam Test: {len(test_entities)}")
print(f"Total Entitas Unik dalam Test: {len(test_entity_freq)}")

# Tambah kolom untuk mengetahui apakah text mengandung entitas
train['has_entity'] = train['text'].str.contains(r'\{@.*?@\}', regex=True)
test['has_entity'] = test['text'].str.contains(r'\{@.*?@\}', regex=True)

print(f"\nTrain - Teks dengan entitas: {train['has_entity'].sum()} ({train['has_entity'].sum()/len(train)*100:.1f}%)")
print(f"Train - Teks tanpa entitas: {(~train['has_entity']).sum()} ({(~train['has_entity']).sum()/len(train)*100:.1f}%)")
print(f"\nTest  - Teks dengan entitas: {test['has_entity'].sum()} ({test['has_entity'].sum()/len(test)*100:.1f}%)")
print(f"Test  - Teks tanpa entitas: {(~test['has_entity']).sum()} ({(~test['has_entity']).sum()/len(test)*100:.1f}%)")

# Top 20 entitas paling sering muncul
print(f"\n{'='*70}")
print("Top 20 Entitas Paling Sering Muncul dalam Train:")
print(f"{'='*70}")
for i, (entity, freq) in enumerate(train_entity_freq.most_common(20), 1):
    print(f"{i:2d}. {entity:40s} : {freq:4d} kali")

# Visualisasi
top_entities = train_entity_freq.most_common(15)
entities = [item[0] for item in top_entities]
freqs = [item[1] for item in top_entities]

plt.figure(figsize=(12, 6))
plt.barh(range(len(entities)), freqs, color='coral', alpha=0.7)
plt.yticks(range(len(entities)), entities, fontsize=9)
plt.xlabel('Frekuensi')
plt.title('Top 15 Entitas Paling Sering Muncul (Train Data)')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

# Distribusi entitas per label
print(f"\n{'='*70}")
print("Analisis Entitas per Label:")
print(f"{'='*70}")
entity_per_label = train.groupby('label_name')['has_entity'].apply(lambda x: (x.sum(), len(x)))
for label, (with_entity, total) in entity_per_label.items():
    pct = (with_entity / total * 100)
    print(f"{label:30s}: {with_entity:4d}/{total:4d} ({pct:5.1f}%) memiliki entitas")
```

```python
# 4. ANALISIS KATA BERDASARKAN KELAS (CLASS-SPECIFIC WORD ANALYSIS)
from sklearn.preprocessing import LabelEncoder
import pandas as pd

# Fungsi untuk menganalisis kata per kelas
def analyze_words_per_class(data, label_column, text_column, top_n=15):
    """
    Analisis kata yang paling berpengaruh terhadap setiap kelas
    Menggunakan TF-IDF untuk mengidentifikasi kata karakteristik per kelas
    """
    
    # Dapatkan unique classes
    classes = data[label_column].unique()
    class_word_stats = {}
    
    print("\n" + "="*80)
    print("ANALISIS KATA BERDASARKAN KELAS (Word Importance per Class)")
    print("="*80)
    
    # Untuk setiap kelas, hitung frekuensi kata
    for cls in classes:
        class_data = data[data[label_column] == cls][text_column]
        
        # Vectorizer untuk menghitung frekuensi kata
        vectorizer = CountVectorizer(max_features=500, lowercase=True, stop_words='english')
        X = vectorizer.fit_transform(class_data)
        
        # Hitung frekuensi kata
        word_freq = dict(zip(vectorizer.get_feature_names_out(), X.toarray().sum(axis=0)))
        word_freq_sorted = sorted(word_freq.items(), key=lambda x: x[1], reverse=True)
        
        class_word_stats[cls] = word_freq_sorted[:top_n]
        
        # Tampilkan hasil
        print(f"\n{'='*80}")
        print(f"TOP {top_n} KATA UNTUK KELAS: {cls}")
        print(f"{'='*80}")
        for i, (word, freq) in enumerate(word_freq_sorted[:top_n], 1):
            print(f"{i:2d}. {word:30s} : {freq:6d} kali")
    
    return class_word_stats

# Jalankan analisis per kelas
class_word_stats = analyze_words_per_class(train, 'label_name', 'text', top_n=15)
```

```python
# 6. ANALISIS CHI-SQUARE: KATA MANA YANG PALING SIGNIFIKAN UNTUK SETIAP KELAS
from scipy.stats import chi2_contingency
from sklearn.feature_extraction.text import CountVectorizer

def get_significant_words_per_class(data, label_column, text_column, n_features=500, top_n=10, p_threshold=0.05):
    """
    Menggunakan Chi-Square test untuk menemukan kata yang paling signifikan (berpengaruh) terhadap kelas
    p_threshold: nilai p untuk menentukan signifikansi (default 0.05)
    """
    
    # Vectorizer
    vectorizer = CountVectorizer(max_features=n_features, lowercase=True)
    X = vectorizer.fit_transform(data[text_column])
    
    # Encode label
    le = LabelEncoder()
    y = le.fit_transform(data[label_column])
    
    feature_names = vectorizer.get_feature_names_out()
    classes = le.classes_
    
    # Dictionary untuk menyimpan hasil
    significant_words = {cls: [] for cls in classes}
    
    print("\n" + "="*80)
    print("CHI-SQUARE TEST: KATA PALING SIGNIFIKAN PER KELAS")
    print("="*80)
    
    # Untuk setiap kata, lakukan chi-square test
    for word_idx, word in enumerate(feature_names):
        # Buat contingency table: word presence vs class
        contingency_table = pd.crosstab(
            X.toarray()[:, word_idx] > 0,  # Ada kata atau tidak
            y  # Class
        )
        
        # Chi-square test
        chi2, p_value, dof, expected = chi2_contingency(contingency_table)
        
        if p_value < p_threshold:
            # Hitung association strength (cramér's V)
            n = contingency_table.sum().sum()
            min_dim = min(contingency_table.shape[0] - 1, contingency_table.shape[1] - 1)
            cramers_v = np.sqrt(chi2 / (n * min_dim)) if min_dim > 0 else 0
            
            # Tentukan kelas mana yang paling berpengaruh
            for cls_idx, cls in enumerate(classes):
                # Hitung conditional probability: P(class | word)
                word_present = X.toarray()[:, word_idx] > 0
                class_mask = y == cls_idx
                
                if word_present.sum() > 0:
                    p_class_given_word = (word_present & class_mask).sum() / word_present.sum()
                    p_class = class_mask.sum() / len(y)
                    
                    # Lift: seberapa banyak kata ini meningkatkan probabilitas kelas
                    lift = p_class_given_word / p_class if p_class > 0 else 0
                    
                    if lift > 1:  # Hanya kata yang positif berpengaruh
                        significant_words[cls].append({
                            'word': word,
                            'chi2': chi2,
                            'p_value': p_value,
                            'lift': lift,
                            'cramers_v': cramers_v
                        })
    
    # Sort dan tampilkan
    for cls in classes:
        significant_words[cls] = sorted(
            significant_words[cls], 
            key=lambda x: x['lift'], 
            reverse=True
        )[:top_n]
        
        print(f"\n{'='*80}")
        print(f"TOP {top_n} KATA PALING SIGNIFIKAN UNTUK KELAS: {cls}")
        print(f"{'='*80}")
        header_line = f"{'No.':<4} {'Kata':<25} {'Lift':<10} {'Cramers V':<12} {'P-value':<12}"
        print(header_line)
        print("-" * 80)
        
        for i, item in enumerate(significant_words[cls], 1):
            print(f"{i:<4} {item['word']:<25} {item['lift']:<10.4f} {item['cramers_v']:<12.6f} {item['p_value']:<12.6e}")
    
    return significant_words

# Jalankan analisis chi-square
significant_words_result = get_significant_words_per_class(train, 'label_name', 'text', top_n=10)
```

```python
# 7. VISUALISASI: LIFT SCORES UNTUK SETIAP KELAS
import matplotlib.pyplot as plt

classes = train['label_name'].unique()
num_classes = len(classes)

# Buat subplot dinamis sesuai jumlah kelas
n_rows = (num_classes + 1) // 2
n_cols = min(2, num_classes)
fig, axes = plt.subplots(n_rows, n_cols, figsize=(16, 5 * n_rows))

# Handle kasus ketika hanya 1 subplot
if n_rows == 1 and n_cols == 1:
    axes = np.array([[axes]])
elif n_rows == 1 or n_cols == 1:
    axes = axes.reshape(n_rows, n_cols)

axes_flat = axes.flatten()

for idx, cls in enumerate(classes):
    if idx < len(axes_flat):
        ax = axes_flat[idx]
        
        # Ambil data untuk kelas ini
        sig_words = significant_words_result[cls]
        
        if len(sig_words) > 0:
            words = [item['word'] for item in sig_words]
            lifts = [item['lift'] for item in sig_words]
            
            # Plot
            colors = ['#2ecc71' if l > 1.5 else '#3498db' for l in lifts]
            bars = ax.barh(range(len(words)), lifts, color=colors, alpha=0.8)
            ax.set_yticks(range(len(words)))
            ax.set_yticklabels(words, fontsize=10)
            ax.set_xlabel('Lift Score', fontsize=11)
            ax.set_title(f'Lift Scores - Kelas: {cls}', fontsize=12, fontweight='bold')
            ax.invert_yaxis()
            ax.axvline(x=1, color='red', linestyle='--', linewidth=2, label='Baseline (Lift=1)')
            ax.grid(axis='x', alpha=0.3)
            ax.legend()
        else:
            ax.text(0.5, 0.5, f'Tidak ada kata signifikan\nuntuk kelas {cls}', 
                    ha='center', va='center', fontsize=12)

# Sembunyikan axes yang tidak digunakan
for idx in range(len(classes), len(axes_flat)):
    axes_flat[idx].axis('off')

plt.tight_layout()
plt.show()

print("Visualisasi Lift Scores selesai!")
```

```python
# 8. HEATMAP: CORRELATION KATA DENGAN KELAS
from sklearn.preprocessing import MultiLabelBinarizer
from sklearn.preprocessing import LabelEncoder

# Ambil top 30 kata paling frequent dari seluruh dataset
vectorizer = CountVectorizer(max_features=30, lowercase=True)
X = vectorizer.fit_transform(train['text'])
feature_names = vectorizer.get_feature_names_out()

# Encode labels
le = LabelEncoder()
y_encoded = le.fit_transform(train['label_name'])

# Buat matrix: untuk setiap kata, hitung frekuensinya di setiap kelas
word_class_matrix = pd.DataFrame(0, index=feature_names, columns=le.classes_)

for word_idx, word in enumerate(feature_names):
    for cls_idx, cls in enumerate(le.classes_):
        # Hitung jumlah dokumen dengan kata ini di kelas ini
        class_mask = y_encoded == cls_idx
        word_count = (X.toarray()[class_mask, word_idx] > 0).sum()
        word_class_matrix.loc[word, cls] = word_count

# Normalisasi per kelas (bagi dengan total dokumen di kelas itu)
word_class_normalized = word_class_matrix.copy()
for cls in le.classes_:
    total_docs = (y_encoded == list(le.classes_).index(cls)).sum()
    word_class_normalized[cls] = word_class_normalized[cls] / total_docs * 100

# Buat heatmap
import seaborn as sns

plt.figure(figsize=(10, 12))
sns.heatmap(word_class_normalized, annot=True, fmt='.1f', cmap='YlOrRd', 
            cbar_kws={'label': 'Persentase Dokumen (%)'}, linewidths=0.5)
plt.title('Heatmap: Frekuensi Kata per Kelas (%)', fontsize=14, fontweight='bold')
plt.ylabel('Kata', fontsize=12)
plt.xlabel('Kelas', fontsize=12)
plt.tight_layout()
plt.show()

print("Heatmap selesai!")
```
